```python
from tensorboard.backend.event_processing import event_accumulator
import matplotlib.pyplot as plt
import numpy as np

def get_training_loss_from_tensorboard(path):
    ea = event_accumulator.EventAccumulator(path)
    ea.Reload()  # 读取数据
    scalar_tensors = ea.Scalars('Training Loss')
    
    loss = [tensor.value for tensor in scalar_tensors]

    return loss

loss1 = get_training_loss_from_tensorboard("train_log/2024-06-02-17-34_cnn_baseline/events.out.tfevents.1717320869.llj-workspace2-kp6xm-22229-worker-0.434.0")
loss2 = get_training_loss_from_tensorboard("train_log/2024-06-06-00-34_timesformer_use_lib_process/events.out.tfevents.1717605281.llj-workspace2-mm7m5-7585-worker-0.457.0")
loss3 = get_training_loss_from_tensorboard("train_log/2024-06-06-00-34_vivit_use_lib_process/events.out.tfevents.1717605270.llj-workspace2-bszpw-7529-worker-0.442.0")
# loss4 = get_training_loss_from_tensorboard("train_log/2024-06-04-00-47_videomae_layer3_wo_pretrain/events.out.tfevents.1717433258.llj-workspace2-xdlzs-9375-worker-0.444.0")


# 创建一个图形和子图
plt.figure(figsize=(10, 6))

# 绘制四条线
plt.plot(range(len(loss1)), loss1, label='ViViT', color='b', linestyle='-', linewidth=2)
plt.plot(range(len(loss2)), loss2, label='CNN+LSTM', color='r', linestyle='--', linewidth=2)
plt.plot(range(len(loss3)), loss3, label='Timesformer', color='g', linestyle='-.', linewidth=2)
# plt.plot(range(len(loss4)), loss4, label='VideoMAE', color='m', linestyle=':', linewidth=2)

# 添加标题和标签
plt.title('Training Loss', fontsize=16)
plt.xlabel('Step', fontsize=14)
plt.ylabel('Loss', fontsize=14)

# 显示图例
plt.legend(fontsize=12)

# 添加网格
plt.grid(True, which='both', linestyle='--', linewidth=0.5)

# 美化图形
plt.tight_layout()

# 显示图形
plt.show()
```
